```python
BD["DIF_sex"] = np.where(BD["sex"] == 1, "Masculino", "Femenino")
BD["DIF_rango_edad"] = np.select(
    [((BD["edad"]>=30) & (BD["edad"]<=39)),
     ((BD["edad"]>=40) & (BD["edad"]<=49)),
     ((BD["edad"]>=50) & (BD["edad"]<=59)),
     ((BD["edad"]>=60))
     ],
    ["30-39", "40-49", "50-59", "60 a más"],
    default="Revisar"
)

BD["DIF_Perfil_evaluado"] = BD["DesTipDirectivo"].str.cat(BD["DesPerfil"], sep=" ")
BD["DIF_Mod_Perfil_evaluado"] = (
    BD["ModalidadEducativa"]
    .str.cat(BD["DesTipDirectivo"], sep=" - ")
    .str.cat(BD["DesPerfil"], sep=" - "))
BD["DIF_D1S2"] = np.where(BD["DIF_Perfil_evaluado"]=="Director Con función docente", np.nan, BD["D1S2"])
BD["DIF_D3S10"] = np.where(BD["DIF_Perfil_evaluado"]=="Subdirector Ninguno", np.nan, BD["D3S10"])
BD["DIF_D3S11"] = np.where(BD["DIF_Perfil_evaluado"]=="Subdirector Ninguno", np.nan, BD["D3S11"])
```

```python
columnas_drop = [
    "ID", "tipo_doc", "ff", "ApePaternoEvaluado", "ApeMaternoEvaluado", "NombreEvaluado", "sex",
    "fecha_nac", "IDLista", "IDEvaluacion", "CodLocal", "NombreIE", "area", "Evaluado_anterior",
    "ESTADODEEVALUACIÓN_anterior", "Evaluado_actualizado", "fecha_pub_hab", "edad", "edquin", "grupedad",
    "HABILITADO", "CÓDIGOMODULAR", "MOTIVOOBSERVACIÓN_anterior", "OBSERVACIÓN"
]

BD.drop(columns=columnas_drop, inplace=True)
```

```python
BD_R_DIF = pd.read_excel("/Users/cristianmolina/Desktop/BD_EDDIR_Reg_Excep/Outputs/BD_DIF_EddirIE_2024_Reg.xlsx",
                         dtype = {"Documento":str})
```

```python
BD_R_DIF.groupby(["ModalidadEducativa", "DIF_Perfil_evaluado", "ESTADODEEVALUACIÓN_actualizado"])\
    .size()\
        .reset_index(name="n")\
            .assign(
                Indice = lambda x: x["ModalidadEducativa"] + " " + x["DIF_Perfil_evaluado"]
            )[["Indice", "ESTADODEEVALUACIÓN_actualizado", "n"]]\
                .pivot_table(index="Indice",
                             columns="ESTADODEEVALUACIÓN_actualizado",
                             values="n"
                             )
```

```python
BD_R_DIF.groupby(["ModalidadEducativa", "DIF_Perfil_evaluado", "ESTADODEEVALUACIÓN_actualizado"])\
    .size()\
        .reset_index(name="n").assign(
                Mod_Estado_Final = lambda x: x["ModalidadEducativa"] + " " + x["ESTADODEEVALUACIÓN_actualizado"]
            )[["DIF_Perfil_evaluado", "Mod_Estado_Final", "n"]]\
                .pivot_table(index="DIF_Perfil_evaluado",
                             columns="Mod_Estado_Final",
                             values="n"
                             )\
                                 .reset_index()\
                                     .to_excel("/Users/cristianmolina/Desktop/BD_EDDIR_Reg_Excep/Outputs/Tabla1_revINFG.xlsx",
                                               index=False)
```

```python
BD_R_DIF.groupby(["Region", "ESTADODEEVALUACIÓN_actualizado"]).size()\
    .reset_index(name = "n")\
        .pivot_table(index = "Region",
                     columns="ESTADODEEVALUACIÓN_actualizado",
                     values="n")\
                         .assign(
                             EVALUADOS = lambda x: x["APROBADO"] + x["DESAPROBADO"] + x["PENDIENTE"]
                         ).to_excel("Tabla2_revINFG.xlsx")
```

```python
# Modalidad EBR
variables = ["D1S1", "D1S2", "D1S3",
             "D1S4", "D2S5", "D2S6",
             "D2S7", "D3S8", "D3S9",
             "D3S10", "D3S11"
             ]

filtro = (

    (BD_R_DIF["ModalidadEducativa"] == "EBR") &
    (
        BD_R_DIF["ESTADODEEVALUACIÓN_actualizado"]
        .isin(["APROBADO", "DESAPROBADO", "PENDIENTE"])
    )
)

df_final = pd.concat([
    (
        BD_R_DIF[
            filtro &
            (BD_R_DIF[var].notna()) &
            (BD_R_DIF[var] != 0)
        ]
        .groupby(var)
        .size()
        .reset_index(name="n")
        .rename(columns={var: "categoria"})
        .assign(variable=var)
    )
    for var in variables
], ignore_index=True)

df_final.pivot_table(index="variable",
                     columns="categoria",
                     values="n",
                     fill_value=0)\
                         .reset_index()\
                         .to_excel("EBR_conteo_sd.xlsx",index=False)
```
